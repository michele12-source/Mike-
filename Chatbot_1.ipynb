```python
# Chatbot molto semplice
print("Ciao! Sono il tuo chatbot. Scrivi 'esci' per chiudere.")

while True:
    user_input = input("Tu: ").lower()  # legge quello che scrivi e lo mette in minuscolo

    if user_input == "esci":
        print("Chatbot: Ciao! Alla prossima!")
        break
    elif "ciao" in user_input:
        print("Chatbot: Ciao! Come stai?")
    elif "come stai" in user_input:
        print("Chatbot: Bene, grazie! E tu?")
    elif "nome" in user_input:
        print("Chatbot: Non ho un vero nome, ma puoi chiamarmi Chatino 😄")
    elif "mi serve il tuo aiuto" in user_input:
        print("Chatbot: Certo, di cosa hai bisogno?")
    elif "quanto fa due più due?" in user_input:
        print("Chatbot: Fa quattro")
    else:
        print("Chatbot: Non so cosa rispondere a questo...")
```
