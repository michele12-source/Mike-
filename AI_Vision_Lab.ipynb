```python
import tensorflow as tf
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np
import requests
from PIL import Image, UnidentifiedImageError
from io import BytesIO

# 1. Carichiamo il "Cervello" dell'IA
model = MobileNetV2(weights='imagenet')

def osserva_immagine(url):
    # Scarica l'immagine dall'URL
    try:
        response = requests.get(url)
        response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)

        # Controlla il tipo di contenuto per assicurarti che sia un'immagine
        content_type = response.headers.get('Content-Type', '')
        if 'image' not in content_type:
            print(f"Errore: Il contenuto scaricato dall'URL '{url}' non è un'immagine. Content-Type: {content_type}")
            return

        img = Image.open(BytesIO(response.content)).resize((224, 224))

        # Prepara l'immagine per l'IA
        x = image.img_to_array(img)
        x = np.expand_dims(x, axis=0)
        x = preprocess_input(x)

        # L'IA fa la sua previsione
        preds = model.predict(x)
        print('Risultato:', decode_predictions(preds, top=3)[0])

    except requests.exceptions.RequestException as e:
        print(f"Errore durante il download dell'immagine dall'URL '{url}': {e}")
    except UnidentifiedImageError:
        print(f"Errore: Impossibile identificare l'immagine dal contenuto scaricato dall'URL: {url}. Potrebbe essere un formato non supportato o un'immagine corrotta.")
    except Exception as e:
        print(f"Si è verificato un errore inatteso: {e}")

# Provala! Incolla qui sotto il link di un'immagine di un cane o un oggetto
url_test = "https://www.meme-arsenal.com/memes/129188a834241ac27e7b686311680d21.jpg"
osserva_immagine(url_test)
```

### Carica l'immagine locale

Per caricare un'immagine dal tuo computer, esegui la cella qui sotto. Ti verrà chiesto di selezionare il file.

```python
from google.colab import files

def osserva_immagine_locale(file_path):
    try:
        img = Image.open(file_path).resize((224, 224))
        # Converti l'immagine in RGB se ha 4 canali (RGBA)
        if img.mode == 'RGBA':
            img = img.convert('RGB')

        # Prepara l'immagine per l'IA
        x = image.img_to_array(img)
        x = np.expand_dims(x, axis=0)
        x = preprocess_input(x)

        # L'IA fa la sua previsione
        preds = model.predict(x)
        print('Risultato:', decode_predictions(preds, top=3)[0])

    except FileNotFoundError:
        print(f"Errore: Il file '{file_path}' non è stato trovato.")
    except UnidentifiedImageError:
        print(f"Errore: Impossibile identificare l'immagine dal file '{file_path}'. Potrebbe essere un formato non supportato o un'immagine corrotta.")
    except Exception as e:
        print(f"Si è verificato un errore inatteso durante l'analisi dell'immagine locale: {e}")


# Carica l'immagine
uploaded = files.upload()

# Prendi il primo file caricato (assumiamo che ne venga caricato solo uno)
if uploaded:
    for fn in uploaded.keys():
        print(f'File caricato: {fn}')
        # Puoi salvare il file temporaneamente o passarlo direttamente alla funzione
        # Per semplicità, lo passiamo direttamente alla funzione aprendolo dal percorso temporaneo
        # Colab salva i file caricati nella directory corrente (/content/)
        osserva_immagine_locale(fn)
else:
    print("Nessun file caricato.")
```
