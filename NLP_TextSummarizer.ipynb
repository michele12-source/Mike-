```python
# Installiamo la libreria per l'italiano
!python -m spacy download it_core_news_sm

import spacy
from collections import Counter
from string import punctuation

nlp = spacy.load("it_core_news_sm")

def crea_riassunto(testo, limite=3):
    # L'IA analizza il testo
    doc = nlp(testo)

    # Trova le parole più importanti (non congiunzioni o punteggiatura)
    parole = [token.text.lower() for token in doc if not token.is_stop and not token.is_punct]
    frequenza_parole = Counter(parole)

    # Punteggio delle frasi in base alle parole importanti che contengono
    punteggio_frasi = {}
    for frase in doc.sents:
        for parola in frase:
            if parola.text.lower() in frequenza_parole:
                punteggio_frasi[frase] = punteggio_frasi.get(frase, 0) + frequenza_parole[parole.index(parola.text.lower())]

    # Prendi le frasi migliori
    riassunto = sorted(punteggio_frasi, key=punteggio_frasi.get, reverse=True)[:limite]
    return " ".join([f.text for f in riassunto])

# PROVALO: Incolla un testo qui sotto
testo_da_studiare = """Napoleone Bonaparte è stato un politico e generale francese, fondatore del Primo Impero francese.
Nato in Corsica, ha condotto numerose campagne militari di successo durante le guerre rivoluzionarie francesi.
È considerato uno dei più grandi strateghi della storia. Morì in esilio all'isola di Sant'Elena."""

print("RIASSUNTO PER LO STUDIO:")
print(crea_riassunto(testo_da_studiare))
```
