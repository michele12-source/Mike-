```python
import json
import os

FILE = "archivio.json"

# ---------------------------
# CARICA E SALVA (Le tue funzioni)
# ---------------------------
def carica_archivio():
    if os.path.exists(FILE):
        with open(FILE, "r") as f:
            try:
                return json.load(f)
            except:
                return []
    return []

def salva_archivio(archivio_da_salvare):
    with open(FILE, "w") as f:
        json.dump(archivio_da_salvare, f, indent=4)

# ---------------------------
# INTERFACCIA DI TEST (Per usarlo su Colab)
# ---------------------------

# 1. Carichiamo l'archivio all'inizio
archivio = carica_archivio()

print("--- BENVENUTO NEL TUO ARCHIVIO DIGITALE ---")

# 2. Chiediamo all'utente di inserire qualcosa
materia = input("Inserisci la materia: ")
voto = input("Inserisci il voto: ")

# 3. Creiamo un nuovo elemento (un dizionario)
nuovo_elemento = {"materia": materia, "voto": voto}
archivio.append(nuovo_elemento)

# 4. Salviamo nell'archivio.json
salva_archivio(archivio)

print(f"\n✅ Ho salvato {materia} con voto {voto}!")
print(f"L'archivio completo ora contiene {len(archivio)} elementi.")
print("-" * 40)
print("CONTENUTO ATTUALE:", archivio)
```
