```python
# Script per generare il file app.py del Progetto 1
codice_progetto_1 = """
import streamlit as st

st.set_page_config(page_title="Digital Titans Search", layout="wide")

st.title("🔍 Digital Titans: Search Engine")
st.write("Analisi dei pilastri dell'informatica moderna.")

# Database sintetico
titans = [
    {"nome": "Steve Jobs", "bio": "Focus: Design e User Experience. Ha reso la tecnologia umana.", "tag": "apple, design, iphone"},
        {"nome": "Bill Gates", "bio": "Focus: Software e Accessibilità. Un PC su ogni scrivania.", "tag": "microsoft, windows, pc"},
            {"nome": "Larry Page & Sergey Brin", "bio": "Focus: Algoritmi e Dati. Organizzare l'informazione mondiale.", "tag": "google, search, data"}
            ]

            query = st.text_input("Inserisci una parola chiave (es. Design, PC, Google):")

            if query:
                for t in titans:
                        if query.lower() in t["nome"].lower() or query.lower() in t["bio"].lower() or query.lower() in t["tag"].lower():
                                    st.success(f"**{t['nome']}**")
                                                st.write(t["bio"])
                                                            st.divider()

                                                            st.subheader("Market Impact (Trillions $)")
                                                            st.bar_chart({"Azienda": ["Apple", "Microsoft", "Google"], "Valore": [3.4, 3.1, 2.1]}, x="Azienda", y="Valore")
                                                            """

with open("app_search.py", "w") as f:
    f.write(codice_progetto_1)

    print("File 'app_search.py' creato con successo! Scaricalo dalla cartella a sinistra.")
```
